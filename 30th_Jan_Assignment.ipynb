```python
# 1
marks = int(input("Enter the marks : "))

if marks > 90:
    print("The student's grade is A")
elif marks > 80 and marks <= 90:
    print("The student's grade is B")
elif marks >= 60 and marks <= 80:
    print("The student's grade is C")
elif marks < 60:
    print("The student's grade is D")
else:
    print("The student is fail")
```

5.  A while loop in Python repeatedly executes a block of code as long as a specified condition remains true.

**condition:** This is a boolean expression that is evaluated before each iteration. If it evaluates to True, the code block inside the loop is executed. If it evaluates to False, the loop terminates, and the program continues with the next statement after the loop.


```python
# 5
number = 1
while number <= 5:
    print(number)
    number += 1
```

```python
# 6
# Program to print 3 different patterns using nested while loops

# Pattern 1: Right-angled triangle of stars
print("Pattern 1: Right-angled triangle")
rows = 5
i = 1
while i <= rows:
    j = 1
    while j <= i:
        print("*", end=" ")
        j += 1
    print()
    i += 1

print("\nPattern 2: Inverted right-angled triangle")
# Pattern 2: Inverted right-angled triangle of stars
i = rows
while i >= 1:
    j = 1
    while j <= i:
        print("*", end=" ")
        j += 1
    print()
    i -= 1

print("\nPattern 3: Number pyramid")
# Pattern 3: Number pyramid
i = 1
while i <= rows:
    spaces = rows - i
    while spaces > 0:
        print("  ", end="")
        spaces -= 1

    k = 1
    while k <= i:
        print(i, end=" ")
        k += 1
    print()
    i += 1
```
